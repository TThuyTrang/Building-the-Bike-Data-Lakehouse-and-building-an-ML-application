```python
%sql

-- Card 1: Total Revenue
WITH kpi_data AS (
  SELECT 
    cy.total_revenue AS current_value,
    py.total_revenue AS prior_value,
    ROUND(((cy.total_revenue - py.total_revenue) / py.total_revenue) * 100, 2) AS yoy_pct
  FROM (
    SELECT SUM(`sales_amount`) AS total_revenue
    FROM `workspace`.`gold`.`fact_sales`
    WHERE YEAR(`order_date`) = 2013
  ) cy
  CROSS JOIN (
    SELECT SUM(`sales_amount`) AS total_revenue
    FROM `workspace`.`gold`.`fact_sales`
    WHERE YEAR(`order_date`) = 2012
  ) py
)
SELECT 
  'Total Revenue' AS metric_name,
  current_value,
  yoy_pct,
  CONCAT(CASE WHEN yoy_pct > 0 THEN '↑ +' ELSE '↓ ' END, FORMAT_NUMBER(yoy_pct, '#,##0.0'), '% vs 2012') AS yoy_label
FROM kpi_data
```
