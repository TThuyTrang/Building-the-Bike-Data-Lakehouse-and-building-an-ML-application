```python
%sql

SELECT 
  p.`product_name`,
  p.`category`,
  p.`subcategory`,
  SUM(f.`sales_amount`) AS total_revenue,
  SUM(f.`quantity`) AS total_quantity_sold,
  ROUND(SUM(f.`sales_amount`) / SUM(f.`quantity`), 2) AS avg_price_per_unit,
  COUNT(DISTINCT f.`order_number`) AS order_count
FROM `workspace`.`gold`.`fact_sales` f
JOIN `workspace`.`gold`.`dim_products` p ON f.`product_key` = p.`product_key`
WHERE f.`order_date` IS NOT NULL
  AND p.`product_name` IS NOT NULL
GROUP BY p.`product_name`, p.`category`, p.`subcategory`
ORDER BY total_revenue DESC
LIMIT 10
```
